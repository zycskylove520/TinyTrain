```python
# Copyright (c) 2025 zycskylove520 - MIT License

"""
使用 网页版jupyter notebook 运行以下代码来显示网络结构图
"""

import torch
import os

from torchvista import trace_model

from tinytrain.models.yolo import YOLOCore
from tinytrain import ROOT

os.chdir(ROOT.parent / "examples/cls")  # Tinytrain/examples/cls

yolo = YOLOCore(link_file="link.toml")
model = yolo.get_model()

model.eval()  # 设置为评估模式

# 准备一个示例输入，用于模型转换和推理
dummy_input = torch.randn(1, 3, 28, 28)  # 形状为 [batch_size, channels, height, width]

trace_model(model, dummy_input)

# 运行完需要重启kernel，避免重复注册engine!
```
